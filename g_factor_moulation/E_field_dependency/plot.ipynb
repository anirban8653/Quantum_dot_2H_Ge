```python
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# Plot settings
# ==================================================
plt.rcParams.update({
    "font.size": 14,
    "axes.labelsize": 18,
    "axes.titlesize": 17,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "legend.fontsize": 13,
    "lines.linewidth": 2.2,
    "lines.markersize": 7,
    "axes.linewidth": 1.4,
    "xtick.major.width": 1.3,
    "ytick.major.width": 1.3,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
})

# ==================================================
# Load data
# ==================================================
file_Ey = "gfactor_zz_vs_Ey_N50_L300.dat"
file_Ez = "gfactor_zz_vs_Ez_N50_L300.dat"

data_Ey = np.loadtxt(file_Ey, comments="#")
data_Ez = np.loadtxt(file_Ez, comments="#")

# Columns:
# 0 size(nm)
# 1 E(eV/Angstrom)
# 2 E(V/um)
# 3 Bz(T)
# 4 gap_plus(eV)
# 5 gap_minus(eV)
# 6 gap_avg(eV)
# 7 g_zz

Ey_Vum = data_Ey[:, 2]
Ez_Vum = data_Ez[:, 2]

gap_Ey_microeV = data_Ey[:, 6] * 1e6
gap_Ez_microeV = data_Ez[:, 6] * 1e6

gzz_Ey = np.abs(data_Ey[:, 7])
gzz_Ez = np.abs(data_Ez[:, 7])

# ==================================================
# Single figure with two panels
# ==================================================
fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.6))

# --------------------------------------------------
# Panel (a): gzz vs electric field
# --------------------------------------------------
ax = axes[0]

ax.plot(
    Ey_Vum,
    gzz_Ey,
    "o-",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$E_y$"
)

ax.plot(
    Ez_Vum,
    gzz_Ez,
    "s--",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$E_z$"
)

ax.set_xlabel(r"$E$ (V/$\mu$m)")
ax.set_ylabel(r"$|g_{zz}|$")
ax.set_title(r"(a) $g_{zz}$ tuning")
ax.legend(frameon=False)
ax.grid(alpha=0.25)

# --------------------------------------------------
# Panel (b): Zeeman splitting vs electric field
# --------------------------------------------------
ax = axes[1]

ax.plot(
    Ey_Vum,
    gap_Ey_microeV,
    "o-",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$E_y$"
)

ax.plot(
    Ez_Vum,
    gap_Ez_microeV,
    "s--",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$E_z$"
)

ax.set_xlabel(r"$E$ (V/$\mu$m)")
ax.set_ylabel(r"$\Delta E$ ($\mu$eV)")
ax.set_title(r"(b) Zeeman splitting")
ax.legend(frameon=False)
ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig("gzz_and_gap_vs_Ey_Ez_N50_L300.pdf", dpi=300, bbox_inches="tight")
plt.savefig("gzz_and_gap_vs_Ey_Ez_N50_L300.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# Plot settings
# ==================================================
plt.rcParams.update({
    "font.family": "DejaVu Serif",
    "mathtext.fontset": "dejavuserif",
    "font.size": 14,
    "axes.labelsize": 18,
    "axes.titlesize": 17,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "legend.fontsize": 13,
    "lines.linewidth": 2.2,
    "lines.markersize": 7,
    "axes.linewidth": 1.4,
    "xtick.major.width": 1.3,
    "ytick.major.width": 1.3,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.top": True,
    "ytick.right": True,
})

# ==================================================
# Load data
# ==================================================
data_Ey_xx = np.loadtxt("gfactor_xx_vs_Ey_N100_L300.dat", comments="#")
data_Ez_xx = np.loadtxt("gfactor_xx_vs_Ez_N100_L300.dat", comments="#")

data_Ey_yy = np.loadtxt("gfactor_yy_vs_Ey_N100_L300.dat", comments="#")
data_Ez_yy = np.loadtxt("gfactor_yy_vs_Ez_N100_L300.dat", comments="#")

data_Ey_zz = np.loadtxt("gfactor_zz_vs_Ey_N100_L300.dat", comments="#")
data_Ez_zz = np.loadtxt("gfactor_zz_vs_Ez_N100_L300.dat", comments="#")

# ==================================================
# Extract electric-field axis
# Column 2 = E (V/um), column 7 = g-factor
# ==================================================
Ey_xx = data_Ey_xx[:, 2]
Ez_xx = data_Ez_xx[:, 2]

Ey_yy = data_Ey_yy[:, 2]
Ez_yy = data_Ez_yy[:, 2]

Ey_zz = data_Ey_zz[:, 2]
Ez_zz = data_Ez_zz[:, 2]

gxx_Ey = np.abs(data_Ey_xx[:, 7])
gxx_Ez = np.abs(data_Ez_xx[:, 7])

gyy_Ey = np.abs(data_Ey_yy[:, 7])
gyy_Ez = np.abs(data_Ez_yy[:, 7])

gzz_Ey = np.abs(data_Ey_zz[:, 7])
gzz_Ez = np.abs(data_Ez_zz[:, 7])

# ==================================================
# Relative change in %
# ==================================================
rel_gxx_Ey = 100 * (gxx_Ey - gxx_Ey[0]) / gxx_Ey[0]
rel_gxx_Ez = 100 * (gxx_Ez - gxx_Ez[0]) / gxx_Ez[0]

rel_gyy_Ey = 100 * (gyy_Ey - gyy_Ey[0]) / gyy_Ey[0]
rel_gyy_Ez = 100 * (gyy_Ez - gyy_Ez[0]) / gyy_Ez[0]

rel_gzz_Ey = 100 * (gzz_Ey - gzz_Ey[0]) / gzz_Ey[0]
rel_gzz_Ez = 100 * (gzz_Ez - gzz_Ez[0]) / gzz_Ez[0]

# ==================================================
# Create three-panel plot
# ==================================================
fig, axes = plt.subplots(1, 3, figsize=(15, 3.8), sharex=True)

# -------------------------
# (a) gxx
# -------------------------
ax = axes[0]

ax.plot(
    Ey_xx, rel_gxx_Ey, "o-",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_y$"
)
ax.plot(
    Ez_xx, rel_gxx_Ez, "s--",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_z$"
)

# ax.axhline(0, color="black", linewidth=1.0, alpha=0.5)
ax.set_xlabel(r"$\mathcal{E}$ (V/$\mu$m)")
ax.set_ylabel(r"$\Delta g_{xx}/g_{xx}(0)$ (%)")
ax.legend(frameon=False)
# ax.grid(alpha=0.25)

ax.text(
    -0.25, 0.97, "(a)",
    transform=ax.transAxes,
    fontsize=18,
    
    va="top",
    ha="left"
)

# -------------------------
# (b) gyy
# -------------------------
ax = axes[1]

ax.plot(
    Ey_yy, rel_gyy_Ey, "o-",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_y$"
)
ax.plot(
    Ez_yy, rel_gyy_Ez, "s--",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_z$"
)

# ax.axhline(0, color="black", linewidth=1.0, alpha=0.5)
ax.set_xlabel(r"$\mathcal{E}$ (V/$\mu$m)")
ax.set_ylabel(r"$\Delta g_{yy}/g_{yy}(0)$ (%)")
ax.legend(frameon=False)
# ax.grid(alpha=0.25)

ax.text(
    -0.2, 0.97, "(b)",
    transform=ax.transAxes,
    fontsize=18,
    
    va="top",
    ha="left"
)

# -------------------------
# (c) gzz
# -------------------------
ax = axes[2]

ax.plot(
    Ey_zz, rel_gzz_Ey, "o-",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_y$"
)
ax.plot(
    Ez_zz, rel_gzz_Ez, "s--",
    markerfacecolor="white",
    markeredgewidth=1.8,
    label=r"$\mathcal{E}_z$"
)

# ax.axhline(0, color="black", linewidth=1.0, alpha=0.5)
ax.set_xlabel(r"$\mathcal{E}$ (V/$\mu$m)")
ax.set_ylabel(r"$\Delta g_{zz}/g_{zz}(0)$ (%)")
ax.legend(frameon=False)
# ax.grid(alpha=0.25)

ax.text(
    -0.25, 0.97, "(c)",
    transform=ax.transAxes,
    fontsize=18,
    
    va="top",
    ha="left"
)

plt.tight_layout()

output_name = "relative_gtensor_tuning_Ey_Ez_N100_L300"

plt.savefig(f"{output_name}.pdf", dpi=300, bbox_inches="tight")
plt.savefig(f"{output_name}.png", dpi=300, bbox_inches="tight")
plt.savefig(f"{output_name}.svg", dpi=300, bbox_inches="tight")

plt.show()
```
